# Model and Tokenizer Testing
Testing the email LSTM model and tokenizer

```python
import pickle
import tensorflow as tf
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences
import numpy as np
```

## Load and Test Tokenizer

```python
# Load tokenizer
with open('email_lstm_tokenizer.pkl', 'rb') as f:
    tokenizer = pickle.load(f)

print("Tokenizer loaded successfully!")
print(f"Vocabulary size: {len(tokenizer.word_index)}")
print(f"First 10 words in vocabulary: {list(tokenizer.word_index.keys())[:10]}")
```

## Load and Test Model

```python
# Load model
model = load_model('email_lstm_model.keras')

print("Model loaded successfully!")
print("\nModel Summary:")
model.summary()
```

## Test with Sample Emails

```python
# Test emails
test_emails = [
    "Congratulations! You've won $1000000! Click here to claim your prize now!",
    "Hi, this is a reminder about our meeting tomorrow at 3 PM.",
    "URGENT: Your account will be suspended unless you verify immediately!",
    "Thank you for your purchase. Your order will be delivered soon."
]

# Get max sequence length from model input shape
max_len = model.input_shape[1]
print(f"Model expects sequences of length: {max_len}")

# Preprocess and predict
for i, email in enumerate(test_emails):
    # Tokenize
    sequence = tokenizer.texts_to_sequences([email])
    # Pad sequence
    padded = pad_sequences(sequence, maxlen=max_len)
    # Predict
    prediction = model.predict(padded, verbose=0)
    
    print(f"\nEmail {i+1}: {email[:50]}...")
    print(f"Spam probability: {prediction[0][0]:.4f}")
    print(f"Classification: {'SPAM' if prediction[0][0] > 0.5 else 'HAM'}")
```

## Tokenizer Functionality Test

```python
# Test tokenizer functions
test_text = "Free money! Click now to win!"

print(f"Original text: {test_text}")
print(f"Tokenized: {tokenizer.texts_to_sequences([test_text])}")
print(f"Padded (max_len={max_len}): {pad_sequences(tokenizer.texts_to_sequences([test_text]), maxlen=max_len)}")
```
