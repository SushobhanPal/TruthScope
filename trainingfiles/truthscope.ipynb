```python
from sklearn.model_selection import train_test_split
xtr,xts,ytr,yts=train_test_split(ip,op,test_size=0.2,random_state=13)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.compose import ColumnTransformer
```

```python
tfidf = ColumnTransformer([
    ('text_tfidf', TfidfVectorizer(max_features=3000), 'text'),
    ('keyword_tfidf', TfidfVectorizer(max_features=500), 'keyword'),
    ('location_tfidf', TfidfVectorizer(max_features=500), 'location')
])
```

```python
X_train_vec = tfidf.fit_transform(xtr)
```

```python
X_val_vec = tfidf.transform(xts)
```

```python
yts
```

```python
ytr
```

```python
from keras.models import Sequential
from keras.layers import Dense,Dropout, BatchNormalization
from keras.optimizers import Adam
from keras.regularizers import l2
```

```python
from keras.models import Sequential
from keras.layers import Dense, Dropout, BatchNormalization
from keras.optimizers import Adam
from keras.regularizers import l2

model = Sequential()
model.add(Dense(128, input_dim=3739, activation='relu'))
model.add(BatchNormalization())
model.add(Dropout(0.3))

model.add(Dense(64, activation='relu', kernel_regularizer=l2(0.001)))
model.add(BatchNormalization())
model.add(Dropout(0.3))

model.add(Dense(1, activation='sigmoid'))

model.compile(optimizer=Adam(learning_rate=0.0001),
              loss='binary_crossentropy',
              metrics=['accuracy'])

model.fit(
    X_train_vec.toarray(), ytr,
    validation_data=(X_val_vec.toarray(), yts),
    epochs=20,
    batch_size=64,
)
```
