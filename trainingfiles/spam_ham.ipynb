```python
df = pd.concat([d1, d2], ignore_index=True)

    df = df.sample(frac=1, random_state=13).reset_index(drop=True)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
df['LABEL'] = le.fit_transform(df['LABEL'])  

X = df['TEXT']
y = df['LABEL']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=39,
    stratify=y
)
vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    max_features=5000
)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print("Train shape:", X_train_vec.shape)
print("Test shape:", X_test_vec.shape)
```

```python
from sklearn.utils.class_weight import compute_class_weight

class_weights = compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)

class_weights_dict = {i : w for i, w in enumerate(class_weights)}
print(class_weights_dict)
```

```python
X_train_dense = X_train_vec.toarray()
X_test_dense = X_test_vec.toarray()

print(X_train_dense.shape, X_test_dense.shape)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam

y_train_np = np.array(y_train)
y_test_np = np.array(y_test)

model = Sequential([
    Input(shape=(X_train_dense.shape[1],)),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    X_train_dense, y_train_np,
    validation_data=(X_test_dense, y_test_np),
    epochs=4,
    batch_size=32,
    class_weight=class_weights_dict,
    verbose=1
)
```

```python
from sklearn.metrics import classification_report, confusion_matrix

y_pred_probs = model.predict(X_test_dense)


y_pred = (y_pred_probs >= 0.5).astype(int)

print("Classification Report:")
print(classification_report(y_test_np, y_pred))

cm = confusion_matrix(y_test_np, y_pred)
print("Confusion Matrix:")
print(cm)
```
