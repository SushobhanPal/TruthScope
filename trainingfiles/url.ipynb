```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
url=pd.read_csv(r"D:\LP2\Tarun Tiwari.zip")
url.tail()
```

```python
url.isnull().sum()
```

```python
url.Label.value_counts()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

# 1. Input/Output split
X = url['URL']  
y = url['Label']

# 2. Train/Test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=13,
    stratify=y  # keeps ham/spam ratio the same in train & test
)

# 3. Vectorize text (TF-IDF)
vectorizer = TfidfVectorizer(
    lowercase=True,
    max_features=5000  # can tweak
)
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

print("Train shape:", X_train_vec.shape)
print("Test shape:", X_test_vec.shape)
```

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.layers import BatchNormalization

# Ensure y is a NumPy array
y_train_np = np.array(y_train)
y_test_np = np.array(y_test)

# Build the model
model = Sequential([
    Input(shape=(X_train_vec.shape[1],)),
    Dense(256, activation='relu'),
    BatchNormalization(),
    Dropout(0.4),
    Dense(128, activation='relu'),
    BatchNormalization(),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(1, activation='sigmoid')
])

model.compile(
    optimizer=Adam(learning_rate=0.0005),
    loss='binary_crossentropy',
 
    metrics=['accuracy']
)

# Train with class weights
history = model.fit(
    X_train_vec, y_train_np,
    validation_data=(X_test_vec, y_test_np),
    epochs=7,
    batch_size=64,
    verbose=1
)
```

```python
# Check if required variables exist
if 'model' not in locals():
    print("Error: model not found. Please run the model training cell first.")
if 'X_test_vec' not in locals():
    print("Error: X_test_vec not found. Please run the data preprocessing cell first.")
```

```python
from sklearn.metrics import classification_report, confusion_matrix
import numpy as np

# Predict probabilities
y_pred_probs = model.predict(X_test_vec)

# Convert probabilities to binary predictions
y_pred = (y_pred_probs >= 0.5).astype(int)

# Classification report
print("Classification Report:")
print(classification_report(y_test_np, y_pred))

# Confusion matrix
cm = confusion_matrix(y_test_np, y_pred)
print("\nConfusion Matrix:")
print(cm)
```
