```python
from keras.models import load_model
import pickle


model = load_model('best_model03_80.h5')


with open('tfidf_vectorizer03.pkl', 'rb') as f:
    tfidf = pickle.load(f)
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=pd.read_csv(r"C:\Users\shubh\Downloads\nlp-getting-started\train.csv")
x
```

```python
x=x.dropna(subset=['keyword'])
```

```python
x['location'] = x['location'].fillna('unknown')
```

```python
x.drop('id',axis=1,inplace=True)
x
```

```python
ip=x.drop('target',axis=1)
ip.head()
```

```python
op=x.drop(ip,axis=1)
op.head()
```

```python
from sklearn.model_selection import train_test_split
xtr,xts,ytr,yts=train_test_split(ip,op,test_size=0.2,random_state=13)
```

```python
X_train_vec = tfidf.fit_transform(xtr)
```

```python
X_val_vec = tfidf.transform(xts)
```

```python
pred = model.predict(X_val_vec)
pred
```

```python
pred_class = (pred > 0.5).astype(int).reshape(-1)
pred_class
```

```python
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

ac=accuracy_score(pred_class,yts)
re=recall_score(pred_class,yts,average='macro') #macro, micro, weighted reason=> study
pr=precision_score(pred_class,yts,average='macro')
f1=f1_score(pred_class,yts,average='macro')

print('Accuracy:',ac)
print('Recall:',re)
print('Precision:',pr)
print('F1 Score:',f1)
```
